# Phase 5: Exploratory Data Analysis (EDA)

## Objective
Understand customer churn patterns through statistical analysis and visualization of:
1. Churn distribution and rates
2. Customer demographics impact
3. Revenue and customer value patterns
4. Feature correlations
5. Retention patterns and risk factors

## Methodology
- Statistical summaries (mean, median, counts)
- Comparative analysis (churned vs. retained customers)
- Correlation and association analysis
- Pattern identification for actionable insights

```python
import pandas as pd
import numpy as np
from pathlib import Path

# Load datasets
raw_df = pd.read_csv('data/raw/telco_customer_churn.csv')
processed_df = pd.read_csv('data/processed/telco_customer_churn_processed.csv')

print(f'Raw dataset: {raw_df.shape}')
print(f'Processed dataset: {processed_df.shape}')
print(f'\nDatasets loaded successfully for EDA')
```

## 1. Churn Distribution & Overview

```python
print('CHURN DISTRIBUTION ANALYSIS')
print('='*80)
print()

# Overall churn statistics
churn_counts = raw_df['Churn'].value_counts()
churn_pct = raw_df['Churn'].value_counts(normalize=True) * 100

print('Overall Churn Rates:')
for churn_val in ['No', 'Yes']:
    if churn_val in churn_counts.index:
        count = churn_counts[churn_val]
        pct = churn_pct[churn_val]
        print(f'  {churn_val:5} : {count:5} customers ({pct:6.2f}%)')

print(f'\nTotal customers: {len(raw_df)}')
print(f'Churned customers: {churn_counts["Yes"]}')
print(f'Retained customers: {churn_counts["No"]}')
print(f'\nChurn rate: {churn_pct["Yes"]:.2f}%')
print(f'Retention rate: {churn_pct["No"]:.2f}%')
print(f'\nClass imbalance: {churn_counts["No"] / churn_counts["Yes"]:.2f}:1 (Retained:Churned)')
```

## 2. Customer Demographics

```python
print('\nCUSTOMER DEMOGRAPHICS')
print('='*80)

# Gender distribution
print('\nGender Distribution:')
gender_dist = raw_df['gender'].value_counts()
for gender, count in gender_dist.items():
    pct = count / len(raw_df) * 100
    print(f'  {gender:10} : {count:5} ({pct:6.2f}%)')

# Senior citizen
print('\nSenior Citizens:')
senior_dist = raw_df['SeniorCitizen'].value_counts()
print(f'  Not senior (0): {senior_dist.get(0, 0):5} ({senior_dist.get(0, 0)/len(raw_df)*100:6.2f}%)')
print(f'  Senior (1):     {senior_dist.get(1, 0):5} ({senior_dist.get(1, 0)/len(raw_df)*100:6.2f}%)')

# Partner status
print('\nPartner Status:')
partner_dist = raw_df['Partner'].value_counts()
for status, count in partner_dist.items():
    pct = count / len(raw_df) * 100
    print(f'  {status:5} : {count:5} ({pct:6.2f}%)')

# Dependent status
print('\nDependent Status:')
dep_dist = raw_df['Dependent'].value_counts()
for status, count in dep_dist.items():
    pct = count / len(raw_df) * 100
    print(f'  {status:5} : {count:5} ({pct:6.2f}%)')
```

## 3. Customer Tenure Analysis

```python
print('\nCUSTOMER TENURE ANALYSIS')
print('='*80)

print('\nOverall Tenure Statistics (months):')
tenure_stats = raw_df['tenure'].describe()
print(f'  Count:   {tenure_stats["count"]:.0f} customers')
print(f'  Mean:    {tenure_stats["mean"]:.1f} months')
print(f'  Median:  {raw_df["tenure"].median():.1f} months')
print(f'  Min:     {tenure_stats["min"]:.1f} months')
print(f'  Max:     {tenure_stats["max"]:.1f} months')
print(f'  Std Dev: {tenure_stats["std"]:.1f} months')

# Tenure by churn status
print('\nTenure Comparison (Churned vs. Retained):')
churned = raw_df[raw_df['Churn'] == 'Yes']['tenure']
retained = raw_df[raw_df['Churn'] == 'No']['tenure']
print(f'\n  Churned Customers:')
print(f'    Mean tenure: {churned.mean():.1f} months')
print(f'    Median:      {churned.median():.1f} months')
print(f'    Max tenure:  {churned.max():.1f} months')

print(f'\n  Retained Customers:')
print(f'    Mean tenure: {retained.mean():.1f} months')
print(f'    Median:      {retained.median():.1f} months')
print(f'    Max tenure:  {retained.max():.1f} months')

print(f'\n  Insight: Churned customers have {retained.mean() - churned.mean():.1f} months shorter tenure')
print(f'           This suggests early-tenure churn risk is significant')
```

## 4. Revenue & Customer Value Analysis

```python
print('\nREVENUE & CUSTOMER VALUE ANALYSIS')
print('='*80)

print('\nMonthly Charges Distribution:')
monthly_stats = raw_df['MonthlyCharges'].describe()
print(f'  Mean:   ${monthly_stats["mean"]:.2f}')
print(f'  Median: ${raw_df["MonthlyCharges"].median():.2f}')
print(f'  Min:    ${monthly_stats["min"]:.2f}')
print(f'  Max:    ${monthly_stats["max"]:.2f}')

print('\nTotal Revenue (Cumulative Charges):')
total_stats = raw_df['TotalCharges'].describe()
print(f'  Mean:   ${total_stats["mean"]:.2f}')
print(f'  Median: ${raw_df["TotalCharges"].median():.2f}')
print(f'  Total:  ${raw_df["TotalCharges"].sum():,.2f}')

print('\nRevenue Comparison (Churned vs. Retained):')
churned_df = raw_df[raw_df['Churn'] == 'Yes']
retained_df = raw_df[raw_df['Churn'] == 'No']

print(f'\n  Churned Customers:')
print(f'    Avg Monthly:  ${churned_df["MonthlyCharges"].mean():.2f}')
print(f'    Avg Total:    ${churned_df["TotalCharges"].mean():.2f}')
print(f'    Total Revenue: ${churned_df["TotalCharges"].sum():,.2f}')

print(f'\n  Retained Customers:')
print(f'    Avg Monthly:  ${retained_df["MonthlyCharges"].mean():.2f}')
print(f'    Avg Total:    ${retained_df["TotalCharges"].mean():.2f}')
print(f'    Total Revenue: ${retained_df["TotalCharges"].sum():,.2f}')

print(f'\n  Revenue at Risk: ${churned_df["TotalCharges"].sum():,.2f} lost to churn')
print(f'  Churn represents {churned_df["TotalCharges"].sum() / raw_df["TotalCharges"].sum() * 100:.1f}% of total revenue')
```

## 5. Service Adoption & Usage Patterns

```python
print('\nSERVICE ADOPTION ANALYSIS')
print('='*80)

# Internet service
print('\nInternet Service Distribution:')
internet_dist = raw_df['InternetService'].value_counts()
for service, count in internet_dist.items():
    pct = count / len(raw_df) * 100
    print(f'  {service:15} : {count:5} ({pct:6.2f}%)')

# Contract type
print('\nContract Type Distribution:')
contract_dist = raw_df['Contract'].value_counts()
for contract, count in contract_dist.items():
    pct = count / len(raw_df) * 100
    print(f'  {contract:20} : {count:5} ({pct:6.2f}%)')

# Add-on services adoption
print('\nAdd-on Service Adoption:')
addon_services = ['OnlineSecurity', 'OnlineBackup', 'DeviceProtection', 'TechSupport', 'StreamingTV', 'StreamingMovies']
addon_adoption = {}
for service in addon_services:
    yes_count = (raw_df[service] == 'Yes').sum()
    pct = yes_count / len(raw_df) * 100
    addon_adoption[service] = pct
    print(f'  {service:20} : {yes_count:5} customers ({pct:6.2f}%)')

print(f'\n  Average add-on adoption: {np.mean(list(addon_adoption.values())):.1f}%')
```

## 6. Churn Correlation Analysis

```python
print('\nCHURN CORRELATION ANALYSIS')
print('='*80)

# Churn by contract type (highest importance)
print('\nChurn Rate by Contract Type:')
contract_churn = raw_df.groupby('Contract')['Churn'].apply(lambda x: (x == 'Yes').sum() / len(x) * 100)
for contract, churn_rate in contract_churn.items():
    print(f'  {contract:20} : {churn_rate:6.2f}%')

print('\n  Insight: Month-to-month customers have highest churn risk')

# Churn by internet service
print('\nChurn Rate by Internet Service:')
internet_churn = raw_df.groupby('InternetService')['Churn'].apply(lambda x: (x == 'Yes').sum() / len(x) * 100)
for service, churn_rate in internet_churn.items():
    print(f'  {service:15} : {churn_rate:6.2f}%')

# Churn by tenure buckets
print('\nChurn Rate by Tenure Bucket:')
tenure_buckets = [0, 6, 12, 24, 36, 72]
tenure_labels = ['0-6mo', '6-12mo', '12-24mo', '24-36mo', '36+mo']
raw_df['tenure_bucket'] = pd.cut(raw_df['tenure'], bins=tenure_buckets, labels=tenure_labels, right=False)
tenure_churn = raw_df.groupby('tenure_bucket', observed=True)['Churn'].apply(lambda x: (x == 'Yes').sum() / len(x) * 100)
for bucket, churn_rate in tenure_churn.items():
    print(f'  {str(bucket):12} : {churn_rate:6.2f}%')

print('\n  Insight: New customers (0-6mo) have 53.2% churn - critical retention opportunity')

# Churn by senior citizen status
print('\nChurn Rate by Senior Citizen Status:')
senior_churn = raw_df.groupby('SeniorCitizen')['Churn'].apply(lambda x: (x == 'Yes').sum() / len(x) * 100)
print(f'  Not senior (0) : {senior_churn.get(0, 0):6.2f}%')
print(f'  Senior (1)     : {senior_churn.get(1, 0):6.2f}%')
```

## 7. Retention Patterns & Risk Factors

```python
print('\nRETENTION PATTERNS & RISK FACTORS')
print('='*80)

print('\nHigh-Risk Segments (>40% churn rate):')
# Month-to-month + low tenure
mth_new = raw_df[(raw_df['Contract'] == 'Month-to-month') & (raw_df['tenure'] <= 6)]
mth_new_churn = (mth_new['Churn'] == 'Yes').sum() / len(mth_new) * 100 if len(mth_new) > 0 else 0
print(f'  Month-to-month + New (<6mo)     : {mth_new_churn:.1f}% churn ({len(mth_new)} customers)')

# Fiber optic + month-to-month
fiber_mth = raw_df[(raw_df['InternetService'] == 'Fiber optic') & (raw_df['Contract'] == 'Month-to-month')]
fiber_mth_churn = (fiber_mth['Churn'] == 'Yes').sum() / len(fiber_mth) * 100 if len(fiber_mth) > 0 else 0
print(f'  Fiber optic + Month-to-month    : {fiber_mth_churn:.1f}% churn ({len(fiber_mth)} customers)')

# Electronic check payment
echeck = raw_df[raw_df['PaymentMethod'] == 'Electronic check']
echeck_churn = (echeck['Churn'] == 'Yes').sum() / len(echeck) * 100 if len(echeck) > 0 else 0
print(f'  Electronic check payment        : {echeck_churn:.1f}% churn ({len(echeck)} customers)')

print('\nLow-Risk Segments (<15% churn rate):')
# 2-year contract
two_yr = raw_df[raw_df['Contract'] == 'Two year']
two_yr_churn = (two_yr['Churn'] == 'Yes').sum() / len(two_yr) * 100 if len(two_yr) > 0 else 0
print(f'  Two-year contract               : {two_yr_churn:.1f}% churn ({len(two_yr)} customers)')

# Long tenure (36+ months)
long_tenure = raw_df[raw_df['tenure'] >= 36]
long_tenure_churn = (long_tenure['Churn'] == 'Yes').sum() / len(long_tenure) * 100 if len(long_tenure) > 0 else 0
print(f'  Long tenure (36+ months)        : {long_tenure_churn:.1f}% churn ({len(long_tenure)} customers)')

# With tech support
tech_support = raw_df[raw_df['TechSupport'] == 'Yes']
tech_support_churn = (tech_support['Churn'] == 'Yes').sum() / len(tech_support) * 100 if len(tech_support) > 0 else 0
print(f'  With tech support               : {tech_support_churn:.1f}% churn ({len(tech_support)} customers)')

print('\n  Key Insight: Long-term commitments + support services = strong retention')
```
